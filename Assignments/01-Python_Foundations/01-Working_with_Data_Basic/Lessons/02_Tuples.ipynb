```python
# Birthday_Tuple (Immutable - can not change)
birthday_tuple = (12, 6, 2002) #(Month, Days, Year)
print("My birthday is on: ", birthday_tuple)

# Access by unpacking
Month, Days, Year = birthday_tuple
print(f"Birthday: {Month}/{Days}/{Year}")
print("Month: ", Month)
print("Days: ", Days)
print("Year: ", Year)

```
