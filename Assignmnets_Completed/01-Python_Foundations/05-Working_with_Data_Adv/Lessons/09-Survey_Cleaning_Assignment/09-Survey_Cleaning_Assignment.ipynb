```python

import pandas as pd

# Load the messy data
df = pd.read_csv("survey_dirty.csv")
df.head()
```

```python
# 1.- Displaying the info of the data frame
df.info()
```

```python
# 2.- Cleaning up string fields by removing extra spaces and turning empty strings
# into missing values 

import pandas as pd

# Clean whitespace in string columns
string_cols = ["name", "email", "gender", "exercise_freq", "consent_given"]

for col in string_cols:
    df[col] = df[col].astype("string").str.strip()

# Converting blank strings to NaN
df[string_cols] = df[string_cols].replace("", pd.NA)
df.head()
```

```python
# 3.- Handling missing values

# Checking how many values are missing
df.isna().sum()
```

```python
# Fill missing categorical values with by dropping the items
# Justification: I dropped rows with missing required survey fields instead of filling them
# because I did not want to invent a category values that were not actually provided
df = df.dropna(subset=["name", "gender", "exercise_freq", "consent_given"])
df.head()
```

```python
# 4.- Standardizing categorical data

import pandas as pd

df = pd.read_csv("survey_dirty.csv")

string_cols = ["name", "email", "gender", "exercise_freq", "consent_given"]
for col in string_cols:
    df[col] = df[col].astype("string").str.strip()

df[string_cols] = df[string_cols].replace("", pd.NA)
df = df.dropna(subset=["name", "gender", "exercise_freq", "consent_given"])

# Converting gender values to consistent categories
gender_map = {
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female",
    "woman": "Female",
    "nonbinary": "Other",
    "other": "Other"
}

df["gender"] = df["gender"].str.lower().map(gender_map)

# Normalizing exercise frequency into standard categories
exercise_map = {
    "daily": "Daily",
    "every day": "Daily",
    "weekly": "Weekly",
    "once a week": "Weekly",
    "3x/week": "Weekly",
    "3x weekly": "Weekly",
    "2 times/week": "Weekly",
    "1-2 times": "Weekly",
    "sometimes": "Rarely",
    "rarely": "Rarely",
    "never": "Never"
}

df["exercise_freq"] = df["exercise_freq"].str.lower().map(exercise_map)

df[["gender", "exercise_freq"]].drop_duplicates().sort_values(["gender", "exercise_freq"])
```

```python
# 5.- Fixing data types

age_words = {"thirty": "30"}
df["age"] = df["age"].astype("string").str.strip().replace(age_words)

df["age"] = pd.to_numeric(df["age"].str.extract(r"(\d+)", expand=False), errors="coerce")
df["height_cm"] = pd.to_numeric(df["height_cm"].astype("string").str.extract(r"(\d+)", expand=False), errors="coerce")
df["weight_kg"] = pd.to_numeric(df["weight_kg"].astype("string").str.extract(r"(\d+)", expand=False), errors="coerce")
df["satisfaction"] = pd.to_numeric(df["satisfaction"], errors="coerce")
df["submitted_at"] = pd.to_datetime(df["submitted_at"], errors="coerce")

df[["age", "height_cm", "weight_kg", "satisfaction", "submitted_at"]].head()

```

```python
# 6.- Validating and cleaning emails


email_pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"

df["email_valid"] = df["email"].str.match(email_pattern, na=False)

print("Invalid emails:")
print(df.loc[~df["email_valid"], ["name", "email"]])

df = df[df["email_valid"]].copy()

df[["name", "email", "email_valid"]].head()
```

```python
#7.- Converting consent_given to Boolean


consent_map = {
    "yes": True,
    "y": True,
    "true": True,
    "no": False,
    "n": False,
    "false": False
}

df["consent_given"] = df["consent_given"].str.lower().map(consent_map)

df[["name", "consent_given"]].head()
```

```python
# 8.- Checking for duplicates

import pandas as pd

# Reload and re-clean the data so this cell works by itself
df = pd.read_csv("survey_dirty.csv")

string_cols = ["name", "email", "gender", "exercise_freq", "consent_given"]
for col in string_cols:
    df[col] = df[col].astype("string").str.strip()

df[string_cols] = df[string_cols].replace("", pd.NA)
df = df.dropna(subset=["name", "gender", "exercise_freq", "consent_given"])

# Find duplicate rows and duplicate respondent IDs
print("Duplicate rows:")
print(df[df.duplicated()])

print("\nDuplicate respondent_id values:")
print(df[df.duplicated(subset=["respondent_id"], keep=False)])

# Drop full duplicate rows
df = df.drop_duplicates()

df.head()
```

```python
# 9.- Saving the cleaned data

import pandas as pd

# Reload and re-clean the data so this cell works by itself
df = pd.read_csv("survey_dirty.csv")

string_cols = ["name", "email", "gender", "exercise_freq", "consent_given"]
for col in string_cols:
    df[col] = df[col].astype("string").str.strip()

df[string_cols] = df[string_cols].replace("", pd.NA)
df = df.dropna(subset=["name", "gender", "exercise_freq", "consent_given"])

# Standardize categorical columns
gender_map = {
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female",
    "woman": "Female",
    "nonbinary": "Other",
    "other": "Other"
}
df["gender"] = df["gender"].str.lower().map(gender_map)

exercise_map = {
    "daily": "Daily",
    "every day": "Daily",
    "weekly": "Weekly",
    "once a week": "Weekly",
    "3x/week": "Weekly",
    "3x weekly": "Weekly",
    "2 times/week": "Weekly",
    "1-2 times": "Weekly",
    "sometimes": "Rarely",
    "rarely": "Rarely",
    "never": "Never"
}
df["exercise_freq"] = df["exercise_freq"].str.lower().map(exercise_map)

consent_map = {
    "yes": True,
    "y": True,
    "true": True,
    "no": False,
    "n": False,
    "false": False
}
df["consent_given"] = df["consent_given"].str.lower().map(consent_map)

age_words = {"thirty": "30"}
df["age"] = df["age"].astype("string").str.strip().replace(age_words)

df["age"] = pd.to_numeric(df["age"].str.extract(r"(\d+)", expand=False), errors="coerce")
df["height_cm"] = pd.to_numeric(df["height_cm"].astype("string").str.extract(r"(\d+)", expand=False), errors="coerce")
df["weight_kg"] = pd.to_numeric(df["weight_kg"].astype("string").str.extract(r"(\d+)", expand=False), errors="coerce")
df["satisfaction"] = pd.to_numeric(df["satisfaction"], errors="coerce")
df["submitted_at"] = pd.to_datetime(df["submitted_at"], errors="coerce")

email_pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"
df["email_valid"] = df["email"].str.match(email_pattern, na=False)
df = df[df["email_valid"]].copy()

# Remove duplicate rows before saving
df = df.drop_duplicates()

# Save cleaned dataset
df.to_csv("survey_cleaned.csv", index=False)

print("Cleaned file saved as survey_cleaned.csv")
df.head()
```
