# 🧼 Cleaning Dirty Data (Missing Values & Type Fixes)

## 🔹 LEARNING GOALS:
- Detect and count missing values (`NaN`)
- Fill or drop missing data
- Convert column data types safely
- Understand the difference between `NaN`, `None`, `""`, and type mismatches


### 🧪 1. Load a Messy Dataset

```python
# Modifying the DataFrame and including some dirty data for cleaning practice

import pandas as pd
import numpy as np

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", None],
    "Age": ["25", "thirty", 35, np.nan, "40"],
    "Signup Date": ["2022-01-01", "not a date", "2022/03/01", None, "April 5, 2022"],
    "Score": [95.5, None, 88.0, 92.5, ""],
    "Email": ["alice@email.com", None, "", "david@email.com", "wrong_email"]
}

df = pd.DataFrame(data)
df
```

```python
# Checking for missing values in the DataFrame
print("Missing values per column:")
print(df.isnull().sum())

print("\nEmpty strings per column:")
print((df == "").sum())
```

```python
# Convertng the "Age" column to numeric, coercing errors to NaN
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df
```

```python
# Convert signup date to datetime, coercing errors to NaT
df["Signup Date"] = pd.to_datetime(df["Signup Date"], errors="coerce")
df
```

```python
# Convert Score 
df["Score"] = pd.to_numeric(df["Score"], errors="coerce")
df
```

```python
# Replace empty strings in the "Email" column with NaN
df["Email"] = df["Email"].replace("", np.nan)
df
```

```python
# Fill missing values
print("Final missing values:")
print(df.isnull().sum())

df.info()
print(df.describe(include="all"))
```

### 🧬 4. Data Type Fixes

```python
df.dtypes
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Score"] = pd.to_numeric(df["Score"], errors="coerce")
df["Signup Date"] = pd.to_datetime(df["Signup Date"], errors="coerce")
df.dtypes
```

### 🩹 5. Impute (Fill In) Fixed Missing Values

```python
df["Age"].fillna(df["Age"].median(), inplace=True)
df["Score"].fillna(df["Score"].mean(), inplace=True)
```

```python
df["Signup Date"].fillna(df["Signup Date"].min(), inplace=True)
```

```python
df["Name"].fillna("Unknown", inplace=True)
```

### 🧠 Mini-Challenge

> 🗂 Load `"../../data/survey.csv"` and:
- Identify which columns have missing values
- Use `.isnull().sum()` to get a null report
- Use a mix of `.fillna()`, `.dropna()`, and `pd.to_numeric()` or `pd.to_datetime()` to clean it
- Print a summary with `.info()` and `.describe()`

```python
import pandas as pd
import numpy as np

survey = pd.read_csv("../../data/survey.csv")
survey.head()
```

```python
# Print missing values per column
print("Missing values per column: ")
print(survey.isnull().sum())
```

```python
# Print summary
print("\nDataFrame info: ")
survey.info()
print(survey.describe(include="all"))
```

```python
# Treating empty string like missig values
survey = survey.replace(["", "NA", "N/A", "null", "NULL", "NONE", "none", "Missing", "MISSING"], np.nan)
print(survey.isnull().sum())
```

```python
# Inspecting columns
print(survey.columns.tolist())
```

```python
# Cleaning Up
for col in survey.columns:
    if "age" in col.lower() or "score" in col.lower() or "income" in col.lower():
        survey[col] = pd.to_numeric(survey[col], errors="coerce")
    if "date" in col.lower():
        survey[col] = pd.to_datetime(survey[col], errors="coerce")
```

```python
# Fill in missing values
for col in survey.select_dtypes(include=["number"]).columns:
    survey[col] = survey[col].fillna(survey[col].median())

for col in survey.select_dtypes(include=["object"]).columns:
    survey[col] = survey[col].fillna("Unknown")
```

```python
# Final Clean Report
print("Final missing values: ")
print(survey.isnull().sum())

survey.info()
print(survey.describe(include="all"))
survey.head()
```
